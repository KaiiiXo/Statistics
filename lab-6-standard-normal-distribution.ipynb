## HISTOGRAM <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# HISTOGRAM PLOT
plt.hist(df['Current'], bins=5,
         alpha = 0.8,
         color = '#e849a9ff',
         edgecolor = 'black')

plt.title('Histogram Of Current Values',
          font='Palatino Linotype',
          fontsize=12)

plt.xlabel('Current(mA)')
plt.ylabel('Frequency')
plt.show()
```

```python
# normal distribution curve
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)

# x-axis
x_min = df['Current'].min()
x_max = df['Current'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)

plt.plot(x, p,
         color="pink",
         linewidth=3)

plt.title('Normal Distribution of Current Values',
          font='Palatino Linotype',
          fontsize=12)
plt.xlabel('Current (mA)')
plt.ylabel('Probability Density')
plt.show()
```

```python
plt.hist(df['Current'], bins=5,
         density = True,
         alpha = 0.8,
         color = '#e849a9ff',
         edgecolor = 'black')
plt.title('Normal Distribution of Current Values',
          font='Palatino Linotype',
          fontsize=12)

plt.xlabel('Current(mA)')
plt.ylabel('Frequency')

# normal distribution curve
p = stats.norm.pdf(x, mean, std)
plt.plot(x, p,
         color='blue',
         linewidth=2)

plt.show()
```

## Standard Normal Distribution N(0,1) <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# standard normam distribution N(0,1)
# x-axis

x_max = df['Z-score'].max()
x_min = df['Z-score'].min()
x = np.linspace(x_min,x_max,100)

p = stats.norm.pdf(x, df['Z-score'].mean(), df['Z-score'].std(ddof=1))

plt.plot(x, p,
         color='pink',
         linewidth=3)

plt.title('Standard Normal Distribution of Current Values N(0,1)',
          font='Palatino Linotype',
          fontsize=12)
          
plt.xlabel('Z-score')
plt.ylabel('Probability Density')
plt.show()
```
