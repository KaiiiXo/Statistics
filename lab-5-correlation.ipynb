```python
# importing packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

## IMPORTING " CURRENT-POWER " DATASET <hr style="border:2.5px solid #e849a9ff"></hr>

```python
df = pd.read_csv(r'C:\Users\Ernie Crislyn\OneDrive\Desktop\Statistics\dataset\current-power.csv')
df
```

## SUMMARIZE STATS <hr style="border:2.5px solid #e849a9ff"></hr>

```python
df. describe()
```

## COVARIANCE OF CURRENT AND POWER <hr style="border:2.5px solid #e849a9ff"></hr>

```python
df. cov()
```

## SCATTER PLOT <hr style="border:2.5px solid #e849a9ff"></hr>

```python
x=df['Current']
y=df['Power']

plt.figure()
plt.scatter(x,y)
plt.title('Device Current vs Power')
plt.ylabel('Power (W)')
plt.xlabel('Current (A)')
plt.show()
```

## CORRELATION COEFFICIENT OF CURRENT AND POWER <hr style="border:2.5px solid #e849a9ff"></hr>

```python
df. corr()
```

## SCATTER AND LINE PLOT WITH REGRESSION LINE <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# regression line
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x,y, color="#e849a9ff", alpha=0.2,label='Data point')

plt.plot(x,line, color='blue' ,label=f'Regression line (r={r_value:.3f})')

plt.title('Device Current vs Power Response')
plt.ylabel('Power (W)')
plt.xlabel('Current (A)')
plt.legend()
plt.show()
```

## AIR QUALITY <hr style="border:2.5px solid #e849a9ff"></hr>

```python
##Load Dataset
air = pd.read_csv(r'C:\Users\Ernie Crislyn\OneDrive\Desktop\Statistics\dataset\air-quality.csv')
air
```

## NOx-HUMIDITY <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# regression line
x = air['NOx']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope*x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x,y, color='#e849a9ff', alpha=0.2,label='Data point')

plt.plot(x,line, color='blue' ,label=f'Regression line (r={r_value:.3f})')

plt.title('NOx vs Humidity Correlation')
plt.ylabel('Relative Humidity(%)')
plt.xlabel('NOx (ug/cu.m)')
plt.legend()
plt.show()
```

 ## NO2-HUMIDITY <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# regression line
x = air['NO2']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope*x + intercept

plt.figure(figsize=(7,4))
plt.scatter(
    x,y, color='#e849a9ff', 
    alpha=0.2,
    label='Data point')

plt.plot(
    x,
    line, 
    color='blue' ,
    label=f'Regression line (r={r_value:.3f})')

plt.title('NOx vs Humidity Correlation')
plt.ylabel('Relative Humidity(%)')
plt.xlabel('NOx (ug/cu.m)')
plt.legend()
plt.show()
```

## TEMPERATURE-HUMIDITY <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# regression line
x = air['Temperature']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope*x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x,y, color='#e849a9ff', alpha=0.2,label='Data point')

plt.plot(x,line, color='blue' ,label=f'Regression line (r={r_value:.3f})')

plt.title('Temperature vs Humidity Correlation')
plt.ylabel('Relative Humidity(%)')
plt.xlabel('NOx (ug/cu.m)')
plt.legend()
plt.show()
```
