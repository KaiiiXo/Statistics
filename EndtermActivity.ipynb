# 1-Sample t-Test <hr style="border:2.5px solid #e849a9ff"></hr>

## Air temperature [K]

```python
# shapiro-wilk normality test
w_stat, p_value = stats.shapiro(df['Air temperature [K]'])

print(f"w_stat: {w_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("Non-normal data")
else:
    # fail to reject null hypothesis
    print("Normal data")
```

```python
"""
left-tailed test
------------------
Ho: us = 1283.78
Ha: us < 1283.78 
"""
pop_mean = 1283.78
t_test, p_value = stats.ttest_1samp(df['Air temperature [K]'],pop_mean,
                                    alternative='less')
print(f"t statistic: {t_test:.4f}")
print(f"P-value: {p_value:.4f}")

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("Air temperature too low for milling target..")
else:
    # fail to reject null hypothesis
    print("Air temperature meets milling target.")
    

# cohen's d
sample_mean = df['Air temperature [K]'].mean()
pop_mean = 1283.78
pop_std = 32.50

d = abs((sample_mean - pop_mean)/pop_std)

print(f"Cohen's d: {d:.4f}")

# power of a test
power = TTestPower().power(
    effect_size = -d, # difference must be negative for alternative='smaller'
    nobs = df['Air temperature [K]'].count(),
    alpha = 0.05,
    alternative = "smaller"
)

print(f"Power of the test: {power*100:.2f}%")
```

# 2-Sample t-Test <hr style="border:2.5px solid #e849a9ff"></hr>

## Air & Process Temperature 

```python
df[['Air temperature [K]','Process temperature [K]']].describe()
```

```python
import matplotlib.pyplot as plt

# Create a figure with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot Air Temperature on the left
ax1.boxplot(df['Air temperature [K]'], tick_labels=['Air temperature [K]'])
ax1.set_ylabel('Temperature [K]')
ax1.set_title('Air Temp')

# Plot Process Temperature on the right
ax2.boxplot(df['Process temperature [K]'], tick_labels=['Process temperature [K]'])
ax2.set_title('Process Temp')

# Automatically adjust the scales so the boxes look "thick"
plt.tight_layout()
plt.show()
```

```python
# shapiro-wilk normality test
print("Air temperature [K]:")
w_stat, p_value = stats.shapiro(df['Air temperature [K]'])

print(f"w_stat: {w_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("Non-normal data")
else:
    # fail to reject null hypothesis
    print("Normal data")
    
print("\nProcess temperature [K]:")
w_stat, p_value = stats.shapiro(df['Process temperature [K]'])

print(f"w_stat: {w_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("Non-normal data")
else:
    # fail to reject null hypothesis
    print("Normal data")
```

```python
from scipy import stats

"""
F-Test for equality of variances
--------------------------------
Ho: var_Air temperature [K] = var_Process temperature [K]
Ha: var_Air temperature [K] != var_Process temperature [K]
"""

# Variances
var_air = df['Air temperature [K]'].var(ddof=1)
var_process = df['Process temperature [K]'].var(ddof=1)

# F-statistic (larger variance divided by smaller)
if var_air > var_process:
    f_stat = var_air / var_process
else:
    f_stat = var_process / var_air

print(f'F-statistic = {f_stat:.4f}')

# Degrees of freedom
dof_air = df['Air temperature [K]'].count() - 1
dof_process = df['Process temperature [K]'].count() - 1

# p-value (two-tailed)
p_value = 2 * min(
    stats.f.cdf(f_stat, dof_air, dof_process),
    1 - stats.f.cdf(f_stat, dof_air, dof_process)
)
print(f'p-value = {p_value:.4f}')

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("unequal variances")
else:
    # fail to reject null hypothesis
    print("equal variances")
```

```python
from scipy import stats

"""
2-Sample T-Test for Independent Samples
----------------------------------------
Ho: Air temperature <= Process temperature
Ha: Air temperature > Process temperature
"""

t_stat, p_value = stats.ttest_ind(
    df['Air temperature [K]'],
    df['Process temperature [K]'],
    alternative='greater',
    equal_var=True
)

print(f't_stat = {t_stat:.4f}')
print(f'p_value = {p_value:.4f}')

# verdict
alpha = 0.05
if p_value < alpha:
    print("Air temperature is significantly higher than Process temperature.")
else:
    print("No significant difference between Air and Process temperatures.")
```

```python
from statsmodels.stats.power import TTestPower

"""
Left-tailed test
------------------
Ho: u_air = u_process
Ha: u_air > u_process
"""

# Cohen's d
air_mean = df['Air temperature [K]'].mean()
process_mean = df['Process temperature [K]'].mean()
air_std = df['Air temperature [K]'].std(ddof=1)

d = abs((air_mean - process_mean) / air_std)
print(f"Cohen's d (Air vs Process): {d:.4f}")

# Power of the test
power = TTestPower().power(
    effect_size=-d,          # negative for left-tailed test
    nobs=df['Air temperature [K]'].count(),
    alpha=0.05,
    alternative="smaller"
)

print(f"Power of the test (Air vs Process): {power*100:.2f}%")
```

# Paired t-Test <hr style="border:2.5px solid #e849a9ff"></hr>

```python
# load dataset
df = pd.read_csv(r"dataset/Tool wearl_maintenance.csv",
                 delimiter=",")
df
```

```python
# summary of dataframe
df.info()
```

```python
# summary of statistics
df.describe()
```

```python
# check for outliers
plt.boxplot(df[['Before','After']],tick_labels=df.columns)
plt.title("Tool Wear Maintenance")
plt.show()
```

```python
"""
F-Test for equality of variances
--------------------------------
Ho: var_before = var_after
Ha: var_before != var_after
"""

var_before = df['Before'].var(ddof=1)
var_after = df['After'].var(ddof=1)

# F-statistic
if var_before > var_after:
    f_stat = var_after/var_before
else:
    f_stat = var_before/var_after

print(f'F-statistic = {f_stat:.4f}')

# Degrees of Freedom
dof_before = df['Before'].count()-1
dof_after = df['After'].count()-1

# probability value
p_value = 1 - stats.f.cdf(f_stat, dof_before, dof_after)
print(f'p-value = {p_value:.4f}')

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("unequal variances")
else:
    # fail to reject null hypothesis
    print("equal variances")
```

```python
"""
Paired T-Test (Tool Wear Maintenance)
"""
t_stat, p_value = stats.ttest_rel(df['Before'],df['After'])
print(f't_stat = {t_stat:.4f}')
print(f'p_value = {p_value:.4f}')

# verdict
alpha = 0.05
if p_value < alpha:
    # reject null hypothesis
    print("There is a significant improvement after the maintenance.")
    # fail to reject null hypothesis
else:
    print("There is no significant difference before and after the maintenance.")
```
